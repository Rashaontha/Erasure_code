# Full SD 1.4 experiment on a CUDA GPU
Use a GPU runtime. The script installs dependencies, downloads the 100 COCO pairs and classifier, and resumes completed outputs when rerun. Set a persistent artifact directory before running. Training can take many hours; interrupted optimization steps restart from their last completed checkpoint.

```python
import os, pathlib, subprocess
import torch
assert torch.cuda.is_available(), 'Select a CUDA GPU runtime'
print(torch.cuda.get_device_name(0))
```

## Colab: mount Drive for durable results
On Kaggle, skip the Drive cell and use `/kaggle/working/erasure_artifacts` (download or publish that directory before the session ends).

```python
if 'COLAB_RELEASE_TAG' in os.environ:
    from google.colab import drive
    drive.mount('/content/drive')
    artifact_root = '/content/drive/MyDrive/erasure_code/gpu_full'
else:
    artifact_root = '/kaggle/working/erasure_artifacts'
pathlib.Path(artifact_root).mkdir(parents=True, exist_ok=True)
print(artifact_root)
```

## Get the repository
For unpushed local changes, upload this repository to the runtime or push them before cloning.

```python
repo = pathlib.Path('/content/Erasure_code' if 'COLAB_RELEASE_TAG' in os.environ else '/kaggle/working/Erasure_code')
if not repo.exists():
    subprocess.run(['git', 'clone', 'https://github.com/palaksingh-git/Erasure_code', str(repo)], check=True)
print(repo)
```

## Run or resume
If model access requires authentication, set `HF_TOKEN` in the runtime environment first. The script writes checkpoints and results directly to `artifact_root`.

```python
env = os.environ.copy()
env['ARTIFACT_ROOT'] = artifact_root
subprocess.run(['bash', 'scripts/run_on_gpu.sh'], cwd=repo, env=env, check=True)
```
